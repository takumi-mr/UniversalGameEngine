# Go (9x9) DQN on Google Colab — Universal Game Engine

Universal Game Engine のバックエンド（Bun + gRPC）を **Colab 内で起動**し、
Python (PyTorch) の DQN エージェント（Double DQN + 合法手マスク）が gRPC の `Reset` / `Step` で 9 路盤の囲碁を自己対戦しながら学習します。
Python 側は囲碁のルールを持たず、合法手（自殺手・コウ）・終局判定・Tromp-Taylor 集計はすべてサーバーの `GoRuleset` が担当します。学習済みモデルは Google Drive に保存され、`uge_rl.serve` で実際の対局相手として使えます。

**観測 / 行動**（`GoTensorAdapter`）: 観測は自分視点の盤面 81（自分 = +1 / 相手 = -1）+ 直前に石が置かれる前の盤面 81 + 連続パス数 1 + 自分視点のコミ 1 = 164 要素で、NN 入力では 7 チャンネル × 9 × 9（自分の石 / 相手の石 / 直前の自分の石 / 直前の相手の石 / パス数 / コミ / 定数 1）に展開します。行動は「打つ点の index（0〜80）+ パス（81）」= 82 通りで、合法手以外の Q 値は `-inf` でマスクします。盤のサイズはサーバーの既定（9 路）を使います。

**注意**: DQN は 1 手ごとの TD 学習なので、囲碁のように報酬が終局時（2 連続パス）にしか出ないゲームでは AlphaZero 版（`go_alphazero_colab.ipynb`）より学習が難しいです。弱いうちはなかなかパスしないので、手数上限（`--max-moves`）で引き分け打ち切りにして、対ランダム勝率が上がるかを見てください。

**手順**: ランタイム → 「ランタイムのタイプを変更」で GPU (T4) を選んでから、上から順に実行してください。

## 5. 学習

`--episodes` を増やすほど強くなります。囲碁（9 路）は 1 エピソードが最大 `--max-moves` 手（既定 200）なので、オセロより 1 エピソードあたり数倍時間がかかります。
`--eps-decay-steps` は総ステップ数（≒ エピソード数 × 平均手数）の 1/3 程度を目安にしてください。
途中経過のチェックポイントも `--save-every` ごとに同じパスへ上書き保存され、中断した場合は `--resume {MODEL_PATH}` で再開できます。

````python
EPISODES = 3000  #@param {type:"integer"}
MAX_MOVES = 200  #@param {type:"integer"}
MODEL_PATH = f'{MODEL_DIR}/go_dqn.pt'

!cd apps/ml && python -m uge_rl.train     --game go     --address localhost:50051     --episodes {EPISODES}     --max-moves {MAX_MOVES}     --eps-decay-steps 200000     --buffer-size 200000     --train-every 2     --out {MODEL_PATH}     --eval-every 100 --eval-games 10     --save-every 100 --log-every 10
````

## 6. 評価（ランダムプレイヤーとの対戦）

`--max-moves` を超えた対局は引き分けになります（省略時はチェックポイントの設定を使います）。

````python
!cd apps/ml && python -m uge_rl.evaluate --checkpoint {MODEL_PATH} --address localhost:50051 --games 50
````

## 7. 学習曲線（対ランダム勝率）

````python
import json
import matplotlib.pyplot as plt

meta = json.load(open(MODEL_PATH.replace('.pt', '.json')))
hist = meta.get('eval_history', [])
if hist:
    ep, wr = zip(*hist)
    plt.plot(ep, wr, marker='o')
    plt.axhline(0.5, ls='--', c='gray')
    plt.xlabel('episode'); plt.ylabel('win rate vs random'); plt.ylim(0, 1)
    plt.title(f"Go DQN ({meta['total_steps']} steps)")
    plt.show()
print({k: meta[k] for k in ('game_type', 'arch', 'total_steps', 'train_steps', 'saved_at', 'git_commit')})
````

## 8. 保存されたファイルと、モデルと対局する方法

- `go_dqn.pt` — PyTorch の state_dict + メタ情報（`uge_rl.checkpoint.load_checkpoint` で復元。`format: uge-rl/dqn/v1`）
- `go_dqn.json` — メタ情報のみ（ゲーム種別・観測形状・行動数・学習ステップ数・学習曲線）

ローカルで対局するには、`.pt` を `models/` にダウンロードしてから

```bash
task rl                                      # バックエンド（RL_MODE=true）
cd apps/frontend && bun dev                  # フロントエンド
cd apps/ml && python -m uge_rl.serve --checkpoint ../../models/go_dqn.pt   # モデルのボットサーバー（DQN は Q 値の argmax で打つ）
```

フロントエンドで囲碁（Go）の「カスタムマッチ」→ 相手を **☁️ gRPC External** にして部屋を作ると、`serve` がその席を見つけて打ち始めます。
